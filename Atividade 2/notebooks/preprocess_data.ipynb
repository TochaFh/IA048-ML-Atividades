Script que gera os arquivos em data/preprocessed a partir dos dados originais de UCI_HAR

```python
import numpy as np
from ml2.config import *
import ml2.preprocessing as pp


# Garantindo que as pastas existam
PREPROCESSED_TRAIN_DIR.mkdir(parents=True, exist_ok=True)
PREPROCESSED_TEST_DIR.mkdir(parents=True, exist_ok=True)


print("--- Processando Versão 1: Atributos (561 features) ---")

# Carregamento
X_train_feat = pp.load_csv_X(TRAIN_UCI_HAR_DIR / "X_train.txt")
y_train = pp.load_csv_y(TRAIN_UCI_HAR_DIR / "y_train.txt")

X_test_feat = pp.load_csv_X(TEST_UCI_HAR_DIR / "X_test.txt")
y_test = pp.load_csv_y(TEST_UCI_HAR_DIR / "y_test.txt")

# Normalização

# Calcula mu e sugma a partir do treino
X_train_feat_norm, mu_feat, sigma_feat = pp.apply_standardization(X_train_feat)

# Reutiliza mu e sigma do treino para aplicar a mesma transformação no test
X_test_feat_norm, _, _ = pp.apply_standardization(X_test_feat, mu_feat, sigma_feat)

# Confirmando dimensões
print(f"Treino V1 -> Amostras (entries): {X_train_feat_norm.shape[0]} | Colunas (features): {X_train_feat_norm.shape[1]}")
print(f"Teste  V1 -> Amostras (entries): {X_test_feat_norm.shape[0]} | Colunas (features): {X_test_feat_norm.shape[1]}")

# Salvando a Versão 1
np.save(PREPROCESSED_TRAIN_DIR / "1X_train.npy", X_train_feat_norm)
np.save(PREPROCESSED_TRAIN_DIR / "1y_train.npy", y_train)
np.save(PREPROCESSED_TEST_DIR / "1X_test.npy", X_test_feat_norm)
np.save(PREPROCESSED_TEST_DIR / "1y_test.npy", y_test)
print("Arquivos da versão 1 salvos com sucesso!")



print("\n--- Processando Versão 2: Sinais Inerciais Brutos (Raw Inertial 2D) ---")
# Listando os 6 canais brutos (ACC total e GYR corpo)
sinais_train = pp.list_inertial_files("train")
sinais_test = pp.list_inertial_files("test")

# Carregamento e concatenação (a matriz já nasce achatada: N_amostras x 768)
X_train_raw = pp.load_files_stacking(TRAIN_UCI_HAR_INERTIAL_DIR, sinais_train)
X_test_raw = pp.load_files_stacking(TEST_UCI_HAR_INERTIAL_DIR, sinais_test)

# Normalização
X_train_raw_norm, mu_raw, sigma_raw = pp.apply_standardization(X_train_raw)
X_test_raw_norm, _, _ = pp.apply_standardization(X_test_raw, mu_raw, sigma_raw)

# Confirmando dimensões
print(f"Treino Raw -> Amostras (entries): {X_train_raw_norm.shape[0]} | Colunas (features): {X_train_raw_norm.shape[1]}")
print(f"Teste  Raw -> Amostras (entries): {X_test_raw_norm.shape[0]} | Colunas (features): {X_test_raw_norm.shape[1]}")

# Salvando a Versão 2 (reaproveitando y_train e y_test)
np.save(PREPROCESSED_TRAIN_DIR / "raw_inertial_X_train.npy", X_train_raw_norm)
np.save(PREPROCESSED_TRAIN_DIR / "raw_inertial_y_train.npy", y_train)
np.save(PREPROCESSED_TEST_DIR / "raw_inertial_X_test.npy", X_test_raw_norm)
np.save(PREPROCESSED_TEST_DIR / "raw_inertial_y_test.npy", y_test)
print("Arquivos da versão raw_inertial salvos com sucesso!")
```
