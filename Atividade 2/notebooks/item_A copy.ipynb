```python
import numpy as np
from ml2.config import *
import ml2.logistic_regression as lr
import ml2.evaluation as ev
```

```python
# 1. Carregando dados do Item A
X_train = np.load(PREPROCESSED_TRAIN_DIR / "1X_train.npy")
y_train = np.load(PREPROCESSED_TRAIN_DIR / "1y_train.npy")
X_test = np.load(PREPROCESSED_TEST_DIR / "1X_test.npy")
y_test = np.load(PREPROCESSED_TEST_DIR / "1y_test.npy")

best_model, best_params, cv_results = lr.train_logistic_regression_cv(
    X_train, y_train,
    cv_splits=5,
    c_values=[0.001, 0.01, 0.1, 1.0, 10.0, 100.0],
    l1_ratios=[0.0, 0.25, 0.5, 0.75, 1.0],
    max_iter=500,
    tol=1e-2
)

print(f"Melhores hiperparâmetros encontrados: {best_params}")
```

```python
# 3. Plot do Heatmap da Busca de Hiperparâmetros
fig_heatmap = lr.plot_cv_hyperparameters_heatmap(cv_results, 0.5, 1)
fig_heatmap.show()
```

```python
# 4. Plot das Features Mais Importantes para a Classe 1 (ex: Caminhada)
fig_feats = lr.plot_top_features_importance(best_model, top_n=15, class_index=0)
fig_feats.show()
```

```python
# Nomes das 6 classes do UCI HAR (opcional, para deixar os eixos com os nomes reais)
class_names = ["CAMINHAR", "SUBIR ESCADAS", "DESCER ESCADAS", "SENTADO", "EM PÉ", "DEITADO"]

# Avaliação completa via evaluation.py
metrics = ev.evaluate_model(
    model=best_model,
    X_test=X_test,
    y_test=y_test,
    class_names=class_names,
    title="Matriz de Confusão - Regressão Logística (561 Features)"
)
```
