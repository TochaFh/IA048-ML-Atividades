```python
import numpy as np
from ml2.config import *
import ml2.knn as knn
import ml2.evaluation as ev
```

```python
# 1. Carregando dados
X_train = np.load(PREPROCESSED_TRAIN_DIR / "1X_train.npy")
y_train = np.load(PREPROCESSED_TRAIN_DIR / "1y_train.npy")
X_test = np.load(PREPROCESSED_TEST_DIR / "1X_test.npy")
y_test = np.load(PREPROCESSED_TEST_DIR / "1y_test.npy")

# Supondo que você já carregou 1X_train.npy e 1y_train.npy
best_model_knn, best_params_knn, cv_results_knn = knn.train_knn_cv(
    X_train, 
    y_train,
    cv_splits=5,
    n_neighbors_list=[1, 3, 5, 7, 11, 15, 21, 31],
    weights_list=['uniform', 'distance'],
    p_list=[1, 2]
)

print(f"Melhores hiperparâmetros encontrados: {best_params_knn}")
```

```python
fig_heatmap = knn.plot_knn_grid_heatmap(cv_results_knn, min_v=0.94, max_v=1)
fig_heatmap.show()
```

```python
fig_lines = knn.plot_knn_grid_lines(cv_results_knn, range_y=[0.92, 1])
fig_lines.show()
```

```python
# Nomes das 6 classes do UCI HAR (opcional, para deixar os eixos com os nomes reais)
class_names = ["CAMINHAR", "SUBIR ESCADAS", "DESCER ESCADAS", "SENTADO", "EM PÉ", "DEITADO"]

# Avaliação completa via evaluation.py
metrics = ev.evaluate_model(
    model=best_model_knn,
    X_test=X_test,
    y_test=y_test,
    class_names=class_names,
    title="Matriz de Confusão - kNN (561 Features)"
)
```
